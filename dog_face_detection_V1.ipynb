```python
#Face Cropping Stanford dataset
from ultralytics import YOLO
import cv2, os
from tqdm import tqdm

model = YOLO("dog_face_yolo/rf_v0_199imgs/weights/best.pt")

STANFORD_ROOT = "/content/datasets/stanford_dogs/Images"
OUTPUT_DIR = "/content/face_crops"
os.makedirs(OUTPUT_DIR, exist_ok=True)

MAX_IMAGES = 3000
count = 0

for breed_name in os.listdir(STANFORD_ROOT):
    breed_path = os.path.join(STANFORD_ROOT, breed_name)
    if not os.path.isdir(breed_path):
        continue

    for img_name in tqdm(os.listdir(breed_path), desc=breed_name):
        if count >= MAX_IMAGES:
            break

        img_path = os.path.join(breed_path, img_name)
        if not img_name.lower().endswith(".jpg"):
            continue

        img = cv2.imread(img_path)
        if img is None:
            continue

        results = model(img_path, conf=0.5)
        if len(results[0].boxes) == 0:
            continue

        x1, y1, x2, y2 = results[0].boxes.xyxy[0].cpu().numpy().astype(int)
        crop = img[y1:y2, x1:x2]

        if crop.size == 0:
            continue

        crop = cv2.resize(crop, (224, 224))
        cv2.imwrite(os.path.join(OUTPUT_DIR, f"{count:05d}.jpg"), crop)
        count += 1

print(f"✅ Saved {count} face crops to {OUTPUT_DIR}")
```

```python
#Triplet Loss
import os
import random
from PIL import Image
from torch.utils.data import Dataset
from torchvision import transforms

class DogTripletDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.root_dir = root_dir
        self.transform = transform

        # Map: dog_id -> list of image paths
        self.dog_to_images = {}
        for dog_id in os.listdir(root_dir):
            dog_path = os.path.join(root_dir, dog_id)
            if os.path.isdir(dog_path):
                imgs = [
                    os.path.join(dog_path, f)
                    for f in os.listdir(dog_path)
                    if f.lower().endswith((".jpg", ".png", ".jpeg"))
                ]
                if len(imgs) >= 2:  # minimum for anchor-positive
                    self.dog_to_images[dog_id] = imgs

        self.dog_ids = list(self.dog_to_images.keys())

    def __len__(self):
        return sum(len(v) for v in self.dog_to_images.values())

    def __getitem__(self, idx):
        # 1. Choose anchor dog
        anchor_dog = random.choice(self.dog_ids)
        anchor_imgs = self.dog_to_images[anchor_dog]

        # 2. Choose anchor & positive (same dog)
        anchor_path, positive_path = random.sample(anchor_imgs, 2)

        # 3. Choose negative dog (different dog)
        negative_dog = random.choice(
            [d for d in self.dog_ids if d != anchor_dog]
        )
        negative_path = random.choice(self.dog_to_images[negative_dog])

        # 4. Load images
        anchor = Image.open(anchor_path).convert("RGB")
        positive = Image.open(positive_path).convert("RGB")
        negative = Image.open(negative_path).convert("RGB")

        if self.transform:
            anchor = self.transform(anchor)
            positive = self.transform(positive)
            negative = self.transform(negative)

        return anchor, positive, negative
#Augumentation
transform = transforms.Compose([
    transforms.Resize((160, 160)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5],
                         std=[0.5, 0.5, 0.5])
])
#DataLoader
from torch.utils.data import DataLoader

dataset = DogTripletDataset(
    root_dir="/content/drive/MyDrive/phase-1",
    transform=transform
)

loader = DataLoader(
    dataset,
    batch_size=16,
    shuffle=True,
    num_workers=2
)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import models

class DogFaceEmbeddingNet(nn.Module):
    def __init__(self, embedding_dim=128):
        super().__init__()

        # Load pretrained backbone
        backbone = models.resnet18(pretrained=True)

        # Remove classifier head
        self.backbone = nn.Sequential(*list(backbone.children())[:-1])
        backbone_out = backbone.fc.in_features  # 512

        # Embedding head
        self.embedding = nn.Sequential(
            nn.Linear(backbone_out, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Linear(256, embedding_dim)
        )

    def forward(self, x):
        x = self.backbone(x)
        x = x.view(x.size(0), -1)
        x = self.embedding(x)

        # L2 normalize embeddings (VERY IMPORTANT)
        x = F.normalize(x, p=2, dim=1)
        return x
```

```python
#Triplet Loss
import torch
import torch.nn as nn
import torch.optim as optim

device = "cuda" if torch.cuda.is_available() else "cpu"

model = DogFaceEmbeddingNet(embedding_dim=128).to(device)

criterion = nn.TripletMarginLoss(
    margin=0.5,      # good default for faces
    p=2
)

optimizer = optim.Adam(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-5
)

#Training LOOP
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()
    running_loss = 0.0

    for anchor, positive, negative in loader:
        anchor = anchor.to(device)
        positive = positive.to(device)
        negative = negative.to(device)

        optimizer.zero_grad()

        emb_a = model(anchor)
        emb_p = model(positive)
        emb_n = model(negative)

        loss = criterion(emb_a, emb_p, emb_n)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    return running_loss / len(loader)
num_epochs = 30   # small dataset → more epochs is OK


#Training Script
for epoch in range(num_epochs):
    loss = train_one_epoch(
        model,
        loader,
        criterion,
        optimizer,
        device
    )

    print(f"Epoch [{epoch+1}/{num_epochs}] | Loss: {loss:.4f}")
#Save the model
torch.save(model.state_dict(), "dogface_embedding_phase1.pth")
```

```python
#Similarity Testing Code

import torch
import torch.nn.functional as F
from PIL import Image
from torchvision import transforms

device = "cuda" if torch.cuda.is_available() else "cpu"

model = DogFaceEmbeddingNet(embedding_dim=128).to(device)
model.load_state_dict(torch.load("dogface_embedding_phase1.pth", map_location=device))
model.eval()

transform = transforms.Compose([
    transforms.Resize((160, 160)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5],
                         std=[0.5, 0.5, 0.5])
])

def get_embedding(img_path):
    img = Image.open(img_path).convert("RGB")
    img = transform(img).unsqueeze(0).to(device)

    with torch.no_grad():
        emb = model(img)

    return emb  # shape: [1, 128]

def cosine_similarity(emb1, emb2):
    return F.cosine_similarity(emb1, emb2).item()

img_a = "/content/drive/MyDrive/phase-1/rayalu/rayalu.png"
img_b = "/content/drive/MyDrive/phase-1/rayalu/rayalu-4.png"   # same dog
img_c = "/content/drive/MyDrive/phase-1/siddhu/vj siddhu-4.png"   # different dog

emb_a = get_embedding(img_a)
emb_b = get_embedding(img_b)
emb_c = get_embedding(img_c)

sim_same = cosine_similarity(emb_a, emb_b)
sim_diff = cosine_similarity(emb_a, emb_c)

print(f"Same dog similarity: {sim_same:.3f}")
print(f"Different dog similarity: {sim_diff:.3f}")

THRESHOLD = 0.75  # tune this

def is_same_dog(img1, img2, threshold=THRESHOLD):
    emb1 = get_embedding(img1)
    emb2 = get_embedding(img2)
    sim = cosine_similarity(emb1, emb2)

    return sim, sim >= threshold

score, result = is_same_dog(img_a, img_b)

print("Similarity:", score)
print("Same dog?" , result)
```

```python
import torch
import torch.nn.functional as F
from torchvision import transforms
from PIL import Image
import ipywidgets as widgets
from IPython.display import display, clear_output
import io # Added this import
from ultralytics import YOLO
import numpy as np

#Face Cropping
FACE_CROPPER_PATH = "/content/drive/MyDrive/Dog_face_Detector.v1-dog-face-v0.yolov8/frozen_models/dog_face_cropper.pt"

face_detector = YOLO(FACE_CROPPER_PATH)

def detect_and_crop_face(pil_img):
    """
    Tries to detect dog face using YOLO.
    If detection fails, assumes image is already cropped.
    """
    img_np = np.array(pil_img)

    results = face_detector(img_np, verbose=False)[0]

    if results.boxes is None or len(results.boxes) == 0:
        # Fallback: assume already cropped face
        return pil_img

    # Pick largest bounding box
    boxes = results.boxes.xyxy.cpu().numpy()
    areas = (boxes[:, 2] - boxes[:, 0]) * (boxes[:, 3] - boxes[:, 1])
    best_box = boxes[areas.argmax()]

    x1, y1, x2, y2 = map(int, best_box)
    face_crop = pil_img.crop((x1, y1, x2, y2))

    return face_crop



def pil_to_widget_image(pil_img, size=(200, 200)):
    pil_img = pil_img.resize(size)
    buffer = io.BytesIO()
    pil_img.save(buffer, format="PNG")
    return widgets.Image(
        value=buffer.getvalue(),
        format='png',
        width=size[0],
        height=size[1]
    )

device = "cuda" if torch.cuda.is_available() else "cpu"

model = DogFaceEmbeddingNet(embedding_dim=128).to(device)
model.load_state_dict(torch.load("dogface_embedding_phase1.pth", map_location=device))
model.eval()

transfrom = transforms.Compose([
    transforms.Resize((160, 160)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5],
                         std=[0.5, 0.5, 0.5])
])


def get_embedding_from_pil(img):
    img = transform(img).unsqueeze(0).to(device)
    with torch.no_grad():
        emb = model(img)
    return emb

def cosine_similarity(emb1, emb2):
    return F.cosine_similarity(emb1, emb2).item()


upload1 = widgets.FileUpload(accept='image/*', multiple=False)
upload2 = widgets.FileUpload(accept='image/*', multiple=False)

button = widgets.Button(description="Compare Dogs", button_style="success")
output = widgets.Output()


THRESHOLD = 0.75

def on_button_click(b):
    clear_output(wait=True)
    display(upload1, upload2, button, output)

    if not upload1.value or not upload2.value:
        with output:
            print("Please upload both images.")
        return

    img1_bytes = list(upload1.value.values())[0]['content']
    img2_bytes = list(upload2.value.values())[0]['content']

    img1 = Image.open(io.BytesIO(img1_bytes)).convert("RGB")
    img2 = Image.open(io.BytesIO(img2_bytes)).convert("RGB")

    face1 = detect_and_crop_face(img1)
    face2 = detect_and_crop_face(img2)

    if face1 is None or face2 is None:
        with output:
            print("❌ Dog face not detected in one or both images.")
        return

    emb1 = get_embedding_from_pil(face1)
    emb2 = get_embedding_from_pil(face2)

    sim = cosine_similarity(emb1, emb2)
    result = "SAME DOG 🐶" if sim >= THRESHOLD else "DIFFERENT DOG ❌"

    img_widget_1 = pil_to_widget_image(face1)
    img_widget_2 = pil_to_widget_image(face2)

    with output:
        display(
            widgets.HBox([
                widgets.VBox([widgets.Label("Detected Face 1"), img_widget_1]),
                widgets.VBox([widgets.Label("Detected Face 2"), img_widget_2])
            ])
        )

        print(f"\nSimilarity Score: {sim:.3f}")
        print(f"Decision: {result}")
button.on_click(on_button_click)

display(upload1, upload2, button, output)
```
